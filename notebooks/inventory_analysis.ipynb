```python
df_clean["Inventory Coverage"] = (
    df_clean["Inventory Level"] / df_clean["Demand Forecast"]
)
```

```python
category_sales = (
    df_clean.groupby("Category")["Units Sold"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
category_inventory = (
    df_clean.groupby("Category")["Inventory Level"]
    .mean()
    .sort_values(ascending=False)
)

category_inventory
```

```python
category_coverage = (
    df_clean.groupby("Category")["Inventory Coverage"]
    .median()
    .sort_values(ascending=False)
)

category_coverage
```

```python
store_coverage = (
    df_clean.groupby("Store ID")["Inventory Coverage"]
    .median()
    .sort_values()
)

store_coverage.head(10)
```

```python
low_coverage = df_clean[df_clean["Inventory Coverage"] < 1]

low_coverage.shape
```

```python
low_coverage_by_category = (
    low_coverage.groupby("Category")
    .size()
    .sort_values(ascending=False)
)

low_coverage_by_category
```

```python
low_coverage_pct = (
    df_clean.groupby("Category")["Inventory Coverage"]
    .apply(lambda x: (x < 1).mean() * 100)
    .sort_values(ascending=False)
)

low_coverage_pct
```

```python
product_pressure = (
    low_coverage.groupby("Product ID")
    .size()
    .sort_values(ascending=False)
)

product_pressure.head(10)
```

```python
product_pressure_pct = (
    df_clean.groupby("Product ID")["Inventory Coverage"]
    .apply(lambda x: (x < 1).mean() * 100)
    .sort_values(ascending=False)
)

product_pressure_pct.head(10)
```
